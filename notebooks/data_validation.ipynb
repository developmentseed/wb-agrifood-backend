```python
import json
```

```python
with open("../data/records_v2.0.json", "r") as f: 
    records = json.loads(f.read())
```

```python
len(records)
```

```python
all([tuple(r.keys()) == ('name', 'transcript excerpt', 'description', 'url', 'type') for r in records])
```

```python
all("name" in r.keys() for r in records)
all("transcript exerpt" in r.keys() for r in records)
all("type" in r.keys() for r in records)
all("description" in r.keys() for r in records)
all("url" in r.keys() for r in records)
```

```python
all(r["name"] is not None for r in records)
all(r["transcript excerpt"] is not None for r in records)
all(r["type"] is not None for r in records)
all(r["description"] is not None for r in records)
all(r["url"] is not None for r in records)
```

```python
set(r["type"] for r in records)
```

```python
len(records)
```

```python
cleaned_records = [
    {"id": str(i), **{k:v for k,v in r.items() if k!= "embedding" and v is not None}} for i, r in enumerate(records)
]

with open("records_v2.0_cleaned.json", "w") as f: 
    f.write(json.dumps(cleaned_records))
```
